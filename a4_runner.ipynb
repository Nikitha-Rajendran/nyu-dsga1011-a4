```python
%%script /content/a4-part2/bin/python
import os
import sys
import torch
from types import SimpleNamespace

os.chdir("/content/part2")
sys.path.insert(0, "/content/part2")

from load_data import T5Dataset, normal_collate_fn
from t5_utils import initialize_model

dataset = T5Dataset("data", "train")
index = max(range(len(dataset)), key=lambda i: len(dataset.target_ids[i]))
batch = normal_collate_fn([dataset[index]])

model = initialize_model(SimpleNamespace(finetune=True))
model.train()
device = next(model.parameters()).device

inputs, mask, decoder_inputs, targets, _ = [
    tensor.to(device) for tensor in batch
]

logits = model(
    input_ids=inputs,
    attention_mask=mask,
    decoder_input_ids=decoder_inputs,
).logits

non_pad = targets != 0
loss = torch.nn.functional.cross_entropy(logits[non_pad], targets[non_pad])
assert torch.isfinite(loss).item(), "Loss is not finite"
loss.backward()

print("Device:", device)
print("SQL target length:", targets.shape[1])
print("Loss:", loss.item())
print("Forward and backward passes succeeded")
```

```python
%%script /content/a4-part2/bin/python
import json
from pathlib import Path
from statistics import mean
from transformers import T5TokenizerFast

tokenizer = T5TokenizerFast.from_pretrained("google-t5/t5-small")
data_dir = Path("/content/part2/data")
output_dir = Path("/content/drive/MyDrive/A4/part2")
assert output_dir.is_dir(), "Drive output folder is unavailable"

def token_stats(texts):
    sequences = tokenizer(
        texts,
        add_special_tokens=False,
        padding=False,
        truncation=False,
    )["input_ids"]

    return {
        "mean_token_length": mean(map(len, sequences)),
        "observed_vocabulary_size": len({
            token for sequence in sequences for token in sequence
        }),
    }

results = {
    "counting_convention": (
        "T5 token IDs; excludes added EOS, padding, and decoder-start tokens. "
        "Unknown tokens are counted when present."
    )
}

for split in ("train", "dev"):
    inputs = (data_dir / f"{split}.nl").read_text().splitlines()
    targets = (data_dir / f"{split}.sql").read_text().splitlines()
    assert len(inputs) == len(targets)

    processed_targets = [
        query.replace("<", "SQL_LESS_THAN") for query in targets
    ]

    input_stats = token_stats(inputs)
    results[split] = {
        "number_of_examples": len(inputs),
        "before": {
            "natural_language": input_stats,
            "sql": token_stats(targets),
        },
        "after": {
            "natural_language": input_stats,
            "sql": token_stats(processed_targets),
        },
    }

output_path = output_dir / "q4_statistics.json"
output_path.write_text(json.dumps(results, indent=2))
print(json.dumps(results, indent=2))
print("Saved to:", output_path)
```

```python
%%script /content/a4-part2/bin/python -u
import os
os.chdir("/content/part2")

import torch
import torch.nn.functional as F
from transformers import T5ForConditionalGeneration
from load_data import get_dataloader

checkpoint = (
    "/content/drive/MyDrive/A4/part2_restart/checkpoints/"
    "scr_experiments/scratch_lr3e4/best"
)
model = T5ForConditionalGeneration.from_pretrained(checkpoint).to("cuda")
model.eval()
loader = get_dataloader(8, "dev")
totals = [0.0, 0.0]
tokens = 0

with torch.inference_mode():
    for i, batch in enumerate(loader):
        if i == 10:
            break
        encoder, mask, decoder, targets, _ = [
            value.to("cuda") for value in batch
        ]
        valid = targets != 0
        tokens += valid.sum().item()

        for j in range(2):
            logits = model(
                input_ids=encoder if j == 0 else encoder.flip(0),
                attention_mask=mask if j == 0 else mask.flip(0),
                decoder_input_ids=decoder,
            ).logits
            totals[j] += F.cross_entropy(
                logits[valid], targets[valid], reduction="sum"
            ).item()

print("Correct-question loss:", totals[0] / tokens)
print("Shuffled-question loss:", totals[1] / tokens)
```

```python
%%script /content/a4-part2/bin/python -u
import os
import json
from pathlib import Path

os.chdir("/content/part2")

import torch
from tqdm import tqdm
from transformers import T5ForConditionalGeneration
from load_data import get_dataloader
from utils import save_queries_and_records, compute_metrics

root = Path("/content/drive/MyDrive/A4/part2_restart")
checkpoint = root / "checkpoints/scr_experiments/scratch_lr3e4/best"
output = root / "scratch_lr3e4_beam4_dev"
output.mkdir(exist_ok=True)

sql_path = output / "predictions.sql"
record_path = output / "records.pkl"
assert not sql_path.exists() and not record_path.exists(), (
    "Beam-search outputs already exist; inspect them before rerunning."
)

model = T5ForConditionalGeneration.from_pretrained(checkpoint).to("cuda")
model.eval()
loader = get_dataloader(4, "dev")
tokenizer = loader.dataset.tokenizer
queries = []

with torch.inference_mode():
    for encoder, mask, _, _, start in tqdm(loader):
        generated = model.generate(
            input_ids=encoder.to("cuda"),
            attention_mask=mask.to("cuda"),
            decoder_input_ids=start.to("cuda"),
            max_new_tokens=640,
            num_beams=4,
            do_sample=False,
        )
        queries.extend(
            text.replace("SQL_LESS_THAN", "<").replace("\n", " ").strip()
            for text in tokenizer.batch_decode(
                generated, skip_special_tokens=True
            )
        )

assert len(queries) == 466
save_queries_and_records(queries, str(sql_path), str(record_path))

sql_em, record_em, f1, errors = compute_metrics(
    "data/dev.sql", str(sql_path),
    "records/ground_truth_dev.pkl", str(record_path),
)

results = {
    "checkpoint": str(checkpoint),
    "num_beams": 4,
    "evaluation_batch_size": 4,
    "record_f1": f1,
    "record_em": record_em,
    "sql_em": sql_em,
    "execution_errors": sum(bool(error) for error in errors),
    "examples": len(queries),
}
(output / "metrics.json").write_text(json.dumps(results, indent=2))
print(json.dumps(results, indent=2))
```
